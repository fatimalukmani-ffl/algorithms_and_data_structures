## Question Three: Move All Zeros to End
**Description**  
Given a list of integers, move all zeros to the end while maintaining the relative order of the non-zero elements. 

**Examples**
```python
Input: nums = [0, 1, 0, 3, 12]
Output: [1, 3, 12, 0, 0]
```
```python
Input: nums = [4, 0, 5, 0, 0, 6]
Output: [4, 5, 6, 0, 0, 0]
```


````python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    # TODO
    pass
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    write = 0  # position to place next non-zero
    
    for read in range(len(nums)):
        if nums[read] != 0:
            nums[write], nums[read] = nums[read], nums[write]
            write += 1
            
    return nums
````

````python
from typing import List

def first_duplicate(nums: List[int]) -> int:
    seen = set()
    for n in nums:
        if n in seen:
            return n
        seen.add(n)
    return -1
````


-   Explain why your solution works


````python
A set keeps track of values we have already seen. While scanning left to right, the first time we encounter a value that is already in the set, that value is the first duplicate encountered in the list, so we return it immediately.
````


-   Explain the problem’s time and space complexity


````python
Time: O(n) because we scan the list once and each set lookup/insert is average O(1).

Space: O(n) in the worst case if all numbers are unique (we store them in the set).
````


-   Explain the thinking to an alternative solution (no coding required, but a classmate reading this should be able to code it up based off your text)


````python
Use a dictionary to store the index of the first time each value appears. As you iterate, if you see a value again, record the index of its second occurrence. After the loop, choose the value with the smallest second-occurrence index (that’s the “first duplicate” by appearance). This is also O(n) time, O(n) space, but a bit more bookkeeping.
````
